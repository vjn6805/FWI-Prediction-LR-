```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
dataset=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
dataset.head()
```

```python
dataset.info()
```

## Data Cleaning

```python
## missing values
dataset[dataset.isnull().any(axis=1)]
```


The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1 : "Bejaia Region Dataset"

2 : "Sidi-Bel Abbes Region Dataset"

Add new column with region

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1
df=dataset
```

```python
df.info()
```

```python
## converting float datatype to integer
df[["Region"]]=df[["Region"]].astype(int)
```

```python
df.head()
df.tail()
```

```python
df.isnull().sum()
```

```python
## removing the null values
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.iloc[[122]]
```

```python
## dropping the row of index 122 as it contains same values as the row of index 121
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
##heare some of the columns have spaces btw thme hence removing them
# fixing spaces in columns name
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

### Changes the required columns as integer data type

```python
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

Changing the other columns to float data datatype

```python
objects=[]
for feature in df.columns:
    if df[feature].dtype=='O':
        objects.append(feature)
print(objects)
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
        
```

```python
df.info()
```

```python
df.describe()
```

## Saving the clean dataset into CSV file

```python
df.to_csv('Algerian_forest_fires_dataset_CLEANED.csv',index=False)
```

## Exploratory Data Analysis

```python
# droping not required parameters
df1=df.drop(['day','month','year'],axis=1)
```

```python
df1.head()
```

```python
# Ecoding the categories in classes
df1['Classes']=np.where(df1['Classes'].str.contains('not fire'),0,1)
```

```python
df1.head()
```

```python
df1['Classes'].value_counts()
```

## Visualization

```python
## ploting the density plot for all features
plt.style.use(style='ggplot')
df1.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
# Percentage for Pie Chart
percentage=df1['Classes'].value_counts(normalize=True)*100
print(percentage)
```

```python
# plottin the piechart
classlabel=['Not Fire','Fire']
plt.figure(figsize=(10,10))
plt.pie(percentage,labels=classlabel,autopct='%1.1f%%',shadow=True)
plt.title('Percentage of Classes')
```
